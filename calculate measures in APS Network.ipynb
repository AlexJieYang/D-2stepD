# APS network measures

Edges point from a citing paper to a cited paper. Set the data directory in config.json before execution.


```python
from project_utils import load_table, output_path
from network_metrics import aps_graph, calculate_cd, calculate_2d
import pandas as pd
```

```python
edges = load_table('aps_citations')
nodes = load_table('aps_nodes')
graph = aps_graph(edges, nodes)
```

```python
records = []
for focal in range(graph.vcount()):
    d, cd = calculate_cd(graph, focal)
    records.append((d, calculate_2d(graph, focal), cd))
metrics = pd.DataFrame(records, columns=['D', '2stepD', 'CD'])
metrics.insert(0, 'id', nodes['id'].to_numpy())
metrics['degree'] = graph.degree()
metrics.to_csv(output_path('aps_disruption_metrics.csv'), index=False)
```
